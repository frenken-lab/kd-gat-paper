# SpecEditor — setup & instructions

Interactive diagram editor for `spec.yaml` files. Renders the SvelteFlow canvas in the notebook and writes edits back to the source file.

**Prerequisites (one-time, run from the repo root on the login node):**

```bash
# 1. Build the widget JS bundle
make speceditor

# 2. Register the kernel (if not done yet)
uv add --dev ipykernel
uv run python -m ipykernel install --user \
    --name kd-gat-paper \
    --display-name "kd-gat-paper (uv)"
```

**In OnDemand:** Kernel → Change Kernel → **kd-gat-paper (uv)**

After editing, rebuild the figure to see the change in the paper:

```bash
make figures FIGURE=kd-gat
```

## Widget cheatsheet

| Action | How |
|--------|-----|
| **Open a diagram** | `editor = SpecEditor('path/to/spec.yaml')` then `editor` on its own line |
| **Edit a node** | Click any node → side panel shows label + role inputs → **Apply** |
| **Change role** | Side panel role dropdown: `vgae` / `gat` / `kd` / `neutral` |
| **Add a bridge** | Side panel Bridges section → **+ Add** → pick from/to components |
| **Remove a bridge** | Click **✕** on the bridge row |
| **Write changes to disk** | `editor.save()` — deep-merges patch into spec.yaml |
| **Discard and reload** | `editor.reload()` — re-reads spec.yaml, resets canvas |
| **Inspect pending patch** | `import json; print(json.loads(editor.patch))` |
| **Pan / zoom** | Drag canvas to pan, scroll to zoom |

**Patch format** (what `editor.patch` contains before `save()`):
```json
{
  "components": { "teacher": { "label": "Teacher GAT" } },
  "bridges": [ { "from": "teacher", "to": "student" } ]
}
```
`bridges` always replaces the full list. `components` only includes keys you changed.
